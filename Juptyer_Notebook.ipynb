Updated Protein Structure 
The project originally considred PDB entry 2DFD, but the residue numbering and identity did not match the intended P119W substitution. The structural workflow was therefore changed to PDB entry 1SEV, which contains glyoxysomal malate dehydrogenase.
In the original 1SEV structure, residue 119 was confirmed to be proline (PRO) in both chains A and B. This made it a suitable starting point for modeling the substitution of proline with tryptophan (TRP) at position 119.

This structure represents glyoxysomal malate dehydrogenase, so it should not be described as human mitochondrial MDH2 unless the original research source confirms that identity.

2. Updated Structure Files

The following files are being used for the project:

1SEV.cif — original downloaded structure.

WT_1SEV.pdb — wild-type reference structure.

P119W_1SEV_mutant.pdb — mutant structure with the P119W substitution.

P119W_verify — PyMOL object reloaded from the saved mutant file to verify the mutation.

The original structure was retained as a reference so that the wild-type protein could be compared with the mutant.
3. Verification of the P119W Mutation

The mutant structure was saved as P119W_1SEV_mutant.pdb, reloaded into PyMOL, and checked by examining the alpha-carbon atoms at residue 119.


```python
## Loading the original protein structure
load 1SEV.clif, glycosomal_mdh
## Loaded the original crystal structure from the 1SEV.cif file into PyMOL and assigned it the object name glyoxosomal_mdh

## Checking the original amino acid at residue 119 
iterate glyoxysomal_mdh and name CA and resi 119, print(chain, resi, resn)
## Output: A 119 PRO, B 119 PRO

## Loading a separate wild-type reference structure
load 1SEV.cif, glyoxysomal_mdh_WT
## Command entered to verify residue 119:
iterate glyoxysomal_mdh_WT and name CA and resi 119, print(chain, resi, resn)

## Saving the wild-type reference
cmd.save("WT_1SEV.pdb", "glyoxysomal_mdh_WT")

## Verifying the P119W mutation
iterate glyoxysomal_mdh and name CA and resi 119, print(chain, resi, resn)
## Output: A 119 TRP, B 119 TRP

##Saving the mutant structure
cmd.save("P119W_1SEV_mutant.pdb", "glyoxysomal_mdh")
## File creared: P119W_1SEV_mutant.pdb

## Counting alpha-carbon atoms in the wild-type structure
count_atoms glyoxysomal_mdh_WT and name CA
## output: 724 atoms
## Purpose:Counted the alpha-carbon atoms in the wild-type object. Since a typical amino acid residue has one alpha-carbon atom, this count provides a useful preliminary measure of how much protein structure is present.

## Counting alpha-carbon atoms in the mutant structure
count_atoms P119W_verify and name CA
## Output: 552 atoms
## Interpretation:The mutant object had 552 alpha-carbon atoms, compared with 724 in the wild-type object. This difference indicates that the two objects do not currently contain matching amounts of protein structure. It does not by itself prove that the mutation removed residues; the objects may contain different chains, residue ranges, or structural selections.
```
